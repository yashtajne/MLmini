# Cervical Cancer Risk Factors (UCI)

Loads the locally downloaded **Cervical Cancer (Risk Factors)** dataset (UCI ID `383`) and previews it with `df.head()`.

* 858 patients x 36 **named, documented** columns (unlike the 32-row `lung+cancer` set, where all attributes are anonymous)
* Values are documented in `variables.csv` (name / role / type), e.g. `Smokes: 0/1`, `Age: years`, `Biopsy: 0/1`
* The raw CSV uses `?` for missing values -> read as `NaN` via `na_values="?"`
* `Hinselmann`, `Schiller`, `Citology`, `Biopsy` are the four diagnostic outcomes

## Load the dataset

```python
from pathlib import Path

import pandas as pd

# Folder next to this notebook, as downloaded from UCI
DATA_FILE = Path("cervical+cancer+risk+factors") / "risk_factors_cervical_cancer.csv"

# The CSV already has a real header row -> no `names=` needed
df = pd.read_csv(DATA_FILE, na_values="?")

print(f"Loaded {DATA_FILE.resolve()}")
df.head()
```

## Quick look at the data

```python
print(f"Shape        : {df.shape[0]} rows x {df.shape[1]} columns")
print(f"Missing (NaN) : {int(df.isna().sum().sum())} values in {int((df.isna().sum() > 0).sum())} columns")
print(f"Duplicate rows: {int(df.duplicated().sum())}")
print()
df.info()
```
