# Lung Cancer Dataset (UCI)

Loads the locally downloaded **Lung Cancer** dataset from `lung+cancer/lung-cancer.data` (UCI ID `62`) and previews it with `df.head()`.

* 32 instances, 56 predictive attributes (nominal, values 0-3), 1 class label (3 classes)
* The raw file has **no header row**, and the first column is the class label
* Missing values are written as `?` in the raw file -> read as `NaN` via `na_values`

> **About `lung-cancer.names`:** that file is the dataset *documentation*, not a header row. It only states that the first attribute is the class label and that all predictive attributes are nominal (0-3); it explicitly says the authors *"give no information on the individual variables"*, so there are no real per-column names to use. Hence `Attribute1...Attribute56`.

## Load the dataset

```python
from pathlib import Path

import pandas as pd


DATA_FILE = Path("lung+cancer") / "lung-cancer.data"

columns = ["class"] + [f"Attribute{i}" for i in range(1, 57)]

df = pd.read_csv(DATA_FILE, header=None, names=columns, na_values="?")

print(f"Loaded {DATA_FILE.resolve()}")
df.head()
```

## Quick look at the data

```python
print(f"Shape   : {df.shape[0]} rows x {df.shape[1]} columns")
print(f"Missing : {int(df.isna().sum().sum())} values")
print()
df.info()
```
